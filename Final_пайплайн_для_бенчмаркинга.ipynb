# Предобработка данных

## Базовая предобработка

```python
output_file_name_preprocessing = 'preprocessing_result_' + today + '.xlsx'
output_notebook_name_preprocessing = 'preprocessing_notebook_' + today + '.ipynb'
```

```python
print("Запуск предобработки данных...\n")


_ = pm.execute_notebook(
    '/content/drive/MyDrive/Диплом/DONE_Только_Предобработка_диплом.ipynb',
    output_notebook_name_preprocessing,
    parameters=dict(input_filename='/content/data_1_2.xlsx', # Передаем имя файла
      output_filename = output_file_name_preprocessing),
    log_output=False,
    progress_bar=False
)

print("\n\nПредобработка завершена, обработанный контент лежит в файле", output_file_name_preprocessing)
```

```python
needed_variable = input("Ведите название переменной: ")

print(f"Обработка переменной {needed_variable}!")
```

```python
output_file_name_benchmarking_report = 'benchmarking_report_' + needed_variable + '_' + today + '.pdf'
output_notebook_name_benchmarking_report = 'benchmarking_report__notebook_' + today + '.ipynb'
```

```python
print("Запуск генерации бенчмаркинговой отчетности...\n")

_ = pm.execute_notebook(
    '/content/drive/MyDrive/Диплом/DONE_Benchmarking.ipynb',
    output_notebook_name_benchmarking_report,
    parameters=dict(input_filename = output_file_name_preprocessing,
                    input_variable = needed_variable,
                    output_filename = output_file_name_benchmarking_report),
    log_output=False,
    log_level="ERROR",
    progress_bar=False
)

print("\n\n Генерация бенчмаркинговой отчетности завершена, обработанный контент лежит в файле", output_file_name_benchmarking_report)
```
