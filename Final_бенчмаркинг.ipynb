```python
def df_to_table_with_pvalue_highlight(df: pd.DataFrame, thr=0.05):
    d = df.copy()


    red_mask = (
        (d['Значимость'] == 1) & (d['Абс. разн'] < 0))
    red_rows = red_mask.fillna(False).to_numpy().nonzero()[0]


    green_mask = (
        (d['Значимость'] == 1) & (d['Абс. разн'] > 0))
    green_rows = green_mask.fillna(False).to_numpy().nonzero()[0]


    data = [d.columns.tolist()] + d.astype(str).values.tolist()
    t = Table(data)


    style = [
        ("FONTNAME", (0,0), (-1,-1), "DejaVu"),
        ("FONTSIZE", (0,0), (-1,-1), 9),
        ("GRID", (0,0), (-1,-1), 0.25, colors.grey),
    ]


    for r in red_rows:
        rr = r + 1
        style += [("BACKGROUND", (0, rr), (-1, rr), colors.HexColor("#ffcbd1")),]


    for r in green_rows:
        rr = r + 1
        style += [("BACKGROUND", (0, rr), (-1, rr), colors.HexColor("#cce7c9")),]


    t.setStyle(TableStyle(style))
    return t
```

```python
from reportlab.lib import colors

# Заголовок бенчмарка
title_style = ParagraphStyle(
    'BenchTitle',
    fontName='DejaVu',
    fontSize=12,
    leading=14,
    textColor=colors.toColor("#2c3e50"),
    spaceAfter=4
)


# Подпись среднего значения
subtitle_style = ParagraphStyle(
    'SubTitle',
    fontName='DejaVu',
    fontSize=9,
    textColor=colors.grey,
    spaceAfter=12
)

text_style = ParagraphStyle(
    name="DejaVuText",
    fontName="DejaVu",
    fontSize=10,
    leading=12,
)
```

```python
from datetime import datetime
```

```python
today = str(datetime.now().strftime('%d_%m_%Y'))

output_filename = 'benchmarking_report_' + input_variable + '_' + today + '.pdf'

print(output_filename)
```

```python

doc = SimpleDocTemplate(output_filename, pagesize=A4)

all_dfs = []
elements = []

for bench in benchmarks:
    current_benchmark = bench
    current_base = input_variable
    x = pd.to_numeric(df[bench], errors="coerce")
    results = []

    for age_group, g in df.groupby(current_base, dropna=False):

        #совокупнсость без текущей группы---------------------------
        rest = x.drop(g.index).dropna()
        if len(rest) < 5:
            continue
        overall_mean = rest.mean()

        #текущая группа ---------------------------------------------
        gx = x.loc[g.index].dropna()
        n = gx.count()

        if n < 5:
            # Слишком маленькая группа для Вилкоксона
            continue

        group_mean = gx.mean()
        group_std = gx.std(ddof=1)

        # Вилкоксон -------------------------------------------------
        diffs = gx - overall_mean
        try:
            stat, p_value = stats.wilcoxon(diffs)
        except ValueError:
            # Если все значения равны overall_mean
            stat, p_value = np.nan, 1.0

        # D Коэна --------------------------------------------------
        if group_std == 0:
            cohens_d = 0
        else:
            cohens_d = (group_mean - overall_mean) / group_std

        results.append({
            "age": age_group,
            "mean": round(group_mean, 2),
            "count": int(n),
            "p_value": round(float(p_value), 4),
            "cohens_d": round(cohens_d, 2),
        })

    test_df = pd.DataFrame(results)

    # Информация для отчета-------------------------------------------
    test_df['abs_diff'] = (test_df['mean'] - overall_mean).round(2)
    test_df['perc_diff'] = (
        ((test_df['mean'] - overall_mean) / overall_mean) * 100
    ).round(1)
    test_df['perc_diff'] = test_df['perc_diff'].apply(lambda x: f"{x:+.1f}%")
    test_df['stat_sign'] = (test_df['p_value'] <= 0.05).astype(int)


    test_df = test_df.rename(columns={
        "age": "Группа",
        "mean": "Среднее",
        "count": "N",
        "p_value": "p-знач",
        "cohens_d": "d-Коэна",
        "abs_diff": "Абс. разн",
        "perc_diff": "% разн",
        "stat_sign": "Значимость"
    })

    all_dfs.append(test_df)

    elements.append(Paragraph(f"Бенчмарк: <b>{bench}</b>", title_style))

    elements.append(Paragraph(
        f"Общее среднее по группам: <b>{round(overall_mean, 2)}</b>",
        subtitle_style
    ))

    elements.append(Spacer(1, 12))
    elements.append(df_to_table_with_pvalue_highlight(test_df, thr=0.05))
    elements.append(Spacer(1, 24))

doc.build(elements)
```
