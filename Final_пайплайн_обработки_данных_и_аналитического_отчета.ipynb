# Предобработка данных

## Базовая предобработка

```python
output_file_name_preprocessing = 'preprocessing_result_' + today + '.xlsx'
output_notebook_name_preprocessing = 'preprocessing_notebook_' + today + '.ipynb'
```

```python
print("Запуск предобработки данных...\n")


_ = pm.execute_notebook(
    '/content/drive/MyDrive/Диплом/DONE_Только_Предобработка_диплом.ipynb',
    output_notebook_name_preprocessing,
    parameters=dict(input_filename='/content/data_1_2.xlsx', # Передаем имя файла
      output_filename = output_file_name_preprocessing),
    log_output=False,
    progress_bar=False
)

print("\n\nПредобработка завершена, обработанный контент лежит в файле", output_file_name_preprocessing)
```

## Категоризация комментариев

```python
output_file_name_comment_tagging = 'with_comment_tags' + output_file_name_preprocessing
output_notebook_name_comment_tagging = 'with_comment_tags_notebook_' + today + '.ipynb'
```

```python
url = 'https://drive.google.com/uc?id=1lfzeY-zRJ9xQCmwlC7VqVFRksZpfHDrf'

output = 'LinearSVC_classifier_model_11may.pkl'
gdown.download(url, output, quiet=True)
print("Параметры модели успешно загружены")
```

```python
print("Запуск категоризации комментариев...\n")


_ = pm.execute_notebook(
    '/content/drive/MyDrive/Диплом/DONE_применение_категоризации.ipynb',
    output_notebook_name_comment_tagging,
    parameters=dict(input_filename = output_file_name_preprocessing,
      output_filename = output_file_name_comment_tagging),
    log_output=False,
    progress_bar=False
)

print("\nКатегоризация завершена, обработанный контент лежит в файле", output_file_name_comment_tagging)
```

# Генерация аналитической отчетности

```python
output_file_name_analytical_report = 'analytical_report_' + today + '.pdf'
output_notebook_name_analytical_report = 'analytical_report_notebook_' + today + '.ipynb'
```

```python
print("Запуск генерации аналитической отчетности...\n")

_ = pm.execute_notebook(
    '/content/drive/MyDrive/Диплом/DONE_аналитическая_отчетность.ipynb',
    output_notebook_name_analytical_report,
    parameters=dict(input_filename = output_file_name_comment_tagging,
                    input_filename_dates = '/content/Дата_заполнения_2.xlsx',
                    output_file_name_analytical_report = output_notebook_name_analytical_report),
    log_output=False,
    log_level="ERROR",
    progress_bar=False
)

print("\n\nГенерация аналитической отчетности завершена, обработанный контент лежит в файле", output_file_name_analytical_report)
```
