# Функции для оформления HTML


```python

def show_colored_box(text):
    html_content = f"""
    <div style="
        background-color: #E8E2EE;
        color: black;
        padding: 20px;
        font-family: Courier;
        font-size: 14px;
        border-radius: 10px;
        width=700px;
    ">
        {text}
    </div>
    """
    display(HTML(html_content))
    return html_content

show_colored_box(info_text);
```

```python
def show_custom_table(data):
    html = '<table style="border-collapse: collapse; font-family:Courier; width: 100%; font-size: 14px;">'
    rows = len(data)
    cols = len(data[0])
    for i, row in enumerate(data):
        html += '<tr>'
        for j, cell in enumerate(row):
            # Horizontal border between rows (except last row)
            border_bottom = 'border-bottom: 1px dashed gray;' if i < rows - 1 else ''
            # Vertical borders: visible except in first row
            border_right = ''
            if i != 0 and j < cols - 1:
                border_right = 'border-right: 1px dashed gray;'
            html += f'<td style="padding:10px; {border_bottom} {border_right}">{cell}</td>'
        html += '</tr>'
    html += '</table>'
    display(HTML(html))
    return html

show_custom_table(table_data);
```

```python

def show_aligned_horizontal_bars(bars, label_width=200):
    """
    bars: list of dictionaries with keys:
        'label': str
        'length': int or float (bar width in pixels)
        'color': str (CSS color name or hex)
        'value': number to display at the end
    label_width: width of the label container in pixels
    """
    html = '<div style="font-family: Courier; font-size: 16px;">'
    for bar in bars:
        html += f"""
        <div style="margin-bottom:10px; display:flex; align-items:center;">
            <div style="width:{label_width}px;">{bar['label']}</div>
            <div style="
                height:20px;
                width:{bar['length']}px;
                background-color:{bar['color']};
                margin-right:10px;
            "></div>
            <div>{bar['value']}</div>
        </div>
        """
    html += '</div>'
    display(HTML(html))
    return html


show_aligned_horizontal_bars(bars);
```

```python

def show_emotional_scale(label, description, score, label_color):
    """
    label: str, text above the row
    description: str, text inside the grey arrow box
    score: float, 0-100, indicating the user's value on the line
    """

    # Sizes
    total_scale_width = 350   # total width of the colored line in px
    scale_height = 12
    arrow_size = 12           # size of the arrow
    arrow_gap = 6             # vertical space between arrow tip and line
    segment_gap = 2           # space between color segments
    grey_box_width = 350      # fixed width for grey box
    grey_box_height = 40

    # Segment proportions (percent of total line)
    green_percent = 25
    yellow_percent = 50
    red_percent = 25

    # Calculate segment widths in px
    green_width = total_scale_width * green_percent / 100
    yellow_width = total_scale_width * yellow_percent / 100
    red_width = total_scale_width * red_percent / 100

    # Arrow position in px according to score (0-100)
    arrow_left = (score / 100) * total_scale_width

    # Determine arrow color based on segment
    if arrow_left <= red_width:
        arrow_color = "#F4AFA9"
    elif arrow_left <= red_width + yellow_width:
        arrow_color = "#F9D9A0"
    else:
        arrow_color = "#9FD7B2"

    # Total container height: enough for arrow + gap + line
    container_height = scale_height + arrow_size + arrow_gap*2 + 10

    html = f"""
    <div style="width:794px; margin:0 auto; font-family: Courier; font-size: 14px; margin-bottom:40px;">
        <!-- Label above -->
        <div style="color: {label_color}; font-weight:bold; font-size: 16px; margin-bottom:10px;">{label}</div>

        <!-- Row: grey box + scale -->
        <div style="display:flex; align-items:center; gap:20px;">

            <!-- Fixed grey arrow box -->
            <div style="
                background: #F4F4F4;
                width:{grey_box_width}px;
                height:{grey_box_height}px;
                padding:10px;
                clip-path: polygon(0 0, 95% 0, 100% 50%, 95% 100%, 0 100%);
                color: #555;
                display:flex;
                font-size: 11px;
                align-items:center;
                justify-content:center;
                text-align:left;
                position: relative;
                down: {scale_height+arrow_size+arrow_gap+10}px;
            ">
                {description}
            </div>

            <!-- Colored scale with arrow -->
            <div style="
                position: relative;
                width:{total_scale_width}px;
                height:{container_height}px;
                display:flex;
                align-items:flex-start;
                gap:{segment_gap}px;
            ">

                <!-- Colored segments -->
                <div style="width:{red_width}px; height:{scale_height}px; background-color:#F4AFA9; margin-top:{arrow_size + arrow_gap}px;"></div>
                <div style="width:{yellow_width}px; height:{scale_height}px; background-color:#F9D9A0; margin-top:{arrow_size + arrow_gap}px;"></div>
                <div style="width:{green_width}px; height:{scale_height}px; background-color:#9FD7B2; margin-top:{arrow_size + arrow_gap}px;"></div>



                <!-- Arrow (on top of line) -->
                <div style="
                    position:absolute;
                    left:{arrow_left}px;
                    bottom:{scale_height + arrow_gap + 2 + 10}px;
                    width:0;
                    height:0;
                    border-left:{arrow_size/2}px solid transparent;
                    border-right:{arrow_size/2}px solid transparent;
                    border-top:{arrow_size}px solid {arrow_color};
                    transform: translateX(-50%);
                "></div>

                <!-- Arrow value above arrow -->
                <div style="
                    position:absolute;
                    left:{arrow_left}px;
                    bottom:{scale_height + arrow_size + arrow_gap + 10}px;
                    width:40px;
                    text-align:center;
                    font-weight:bold;
                    color:black;
                    transform: translateX(-50%);
                ">
                    {score}
                </div>
            </div>
        </div>
    </div>
    """
    display(HTML(html))
    return html

for emotional_scale in emotional_scales:
  print(emotional_scale)
  show_emotional_scale(
      label = emotional_scale['label'],
      label_color = emotional_scale['label_color'],
      description = emotional_scale['description'],
      score = emotional_scale['value']
  )
```
